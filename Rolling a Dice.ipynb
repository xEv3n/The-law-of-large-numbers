Objective: Simulate rolling a fair six-sided die 100, 500, 1,000, and 10,000 times, and observe how the empirical probability of rolling a 6 converges toward the theoretical probability (1/6 ≈ 16.67%) as the number of simulations increases.

```python
import random 
import matplotlib.pyplot as plt 
```

```python
simulations = [100,500,1000,10000]
six = []


for simulation in simulations:
    dice_six = 0
    dice_roll = 0
    while dice_roll < simulation:
     throws= random.choice([1,2,3,4,5,6])
     if throws == 6:
        dice_six += 1
     else:
        pass
     dice_roll += 1

    six_percentage = round((dice_six/simulation)*100,2)
    six.append(six_percentage)

```

```python
#Plotting using Matplotlib 

plt.figure(figsize= (10,5))
plt.bar([str(s) for s in simulations], six , color = "deepskyblue")
plt.axhline(y = 16.67, linestyle ="--", color = "red", label = "Expected 16.67%")
plt.xlabel("Number of simulations (sample size)")
plt.ylabel("Percentage of six (%)")
plt.title("Getting a Six(%) vs Sample size")
plt.grid()
plt.legend()
```
