The law of Large number (Using Coin Toss)

1. Run simulations by tossing coin 10, 50, 100, 1,000 and 10,000 times.
2. Calculate and record the number of times "Head" is the outcome and same for the "Tails".
3. Calculate the Percentage of both Heads and Tails by taking respective simulation into consideration.
4. Plot it on the graph 
5. *Objective*: As the number of simulations kept on increasing the chances of getting Heads will come to its expected value. In other words, Higher the population or dataset more reliable it is 
6. Expected outcome: The more times you repeat a coin toss, the closer the average result tends to get to the expected result.


```python
import random  
import matplotlib.pyplot as plt 
```

```python
simulations = [10,50,100,1000,10000]
heads_percentages = []
#Making Simulation a list for loop
for simulation  in simulations:
    Heads = 0 
    Tails = 0 
    coin_toss = 0 

    while coin_toss < simulation:
        choose = random.choice(["Heads","Tails"]) 
        if choose == "Heads":
            Heads +=1
        else:
            Tails += 1
        coin_toss += 1

    #Calculating the percentage of Heads as simulations goes up
    Heads_percentage = round((Heads/simulation)* 100,2)
    heads_percentages.append(Heads_percentage)

#Plotting on chart 

plt.figure(figsize=(10, 5))
plt.bar([str(s) for s in simulations], heads_percentages, color= "skyblue")
plt.axhline(y=50, color='red', linestyle='--', label='Expected (50%)')
plt.xlabel("Number of times coin tosses (sample size)")
plt.ylabel("Percentage of Heads(%)")
plt.title("Heads % vs Sample Size")
plt.legend()
```
